```python
import itertools

def combinaisons_successives(action_dim, nb_repetitions):
    # Générer toutes les combinaisons possibles de taille K avec répétition
    elements = list(range(action_dim))
    combinaisons = list(itertools.product(elements, repeat=nb_repetitions))

    # Convertir chaque combinaison en un tensor PyTorch
    combinaisons_tensor = [torch.tensor(combi, dtype=torch.int64) for combi in combinaisons]

    return combinaisons_tensor

class DiscreteQAgent(Agent):
    def __init__(self, state_dim, hidden_layers, action_dim):
        super().__init__()
        self.model = build_mlp(
            [state_dim] + list(hidden_layers) + [action_dim], activation=nn.ReLU()
        )

    def forward(self, t, **kwargs):
        obs = self.get(("env/env_obs", t))
        q_values = self.model(obs)
        self.set((f"{self.prefix}q_values", t), q_values)

```

```python

class ArgmaxActionSelector(Agent):
    """BBRL agent that selects the best action based on Q(s,a)"""

    def forward(self, t: int, **kwargs):
        q_values = self.get(("q_values", t))

        action = q_values.argmax(-1)

        # Need to transform a tensor of neuron indexes into their corresponding code
        # The code below won't work
        self.set(("action", t), action)

class EGreedyActionSelector(Agent):
    def __init__(self, epsilon):
        super().__init__()
        self.epsilon = epsilon

    def forward(self, t: int, **kwargs):
        # Retrieves the q values
        # (matrix nb. of episodes x nb. of actions)
        q_values: torch.Tensor = self.get(("q_values", t))
        size, nb_actions = q_values.shape

        # Flag
        is_random = torch.rand(size) < self.epsilon

        # Actions (random / argmax)

        # Need to transform a tensor of neuron indexes into their corresponding code
        # The code below won't work
        max_action = q_values.argmax(-1)

        # Same need for random action indexes
        random_action = torch.randint(nb_actions, size=(size,))

        # Choose the action based on the is_random flag
        action = torch.where(is_random, random_action, max_action)

        # Sets the action at time t
        self.set(("action", t), action)
```

```python

def dqn_compute_critic_loss(
    cfg, reward, must_bootstrap, q_values, target_q_values, action
):
    """Compute the critic loss

    :param reward: The reward $r_t$ (shape 2 x B)
    :param must_bootstrap: The must bootstrap flag at $t+1$ (shape 2 x B)
    :param q_values: The Q-values (shape 2 x B x A)
    :param target_q_values: The target Q-values (shape 2 x B x A)
    :param action: The chosen actions (shape 2 x B)
    :return: _description_
    """

    # Implement the DQN loss

    # Adapt from the previous notebook and adapt to our case (target Q network)
    # Don't forget that we deal with transitions (and not episodes)

    # Q(st, at) from q_values[0]
    q_values_for_action = torch.gather(q_values[0], dim=1, index=action[0].unsqueeze(-1)).squeeze(-1)

    max_target_q_values = target_q_values[1].max(dim=1)[0].detach()

    target = reward[1] + cfg.algorithm.discount_factor * must_bootstrap[1] * max_target_q_values

    # MSE loss
    mse = nn.MSELoss()
    critic_loss = mse(q_values_for_action, target)

    return critic_loss
```

```python
class DQN(EpochBasedAlgo):
    def __init__(self, cfg, wrappers=[]):
        super().__init__(cfg, env_wrappers=wrappers)

        obs_size, act_size = self.train_env.get_obs_and_actions_sizes()

        # Get the two agents (critic and target critic)
        critic = DiscreteQAgent(
            obs_size, cfg.algorithm.architecture.hidden_size, act_size
        )
        target_critic = copy.deepcopy(critic).with_prefix("target/")

        # Builds the train agent that will produce transitions
        explorer = EGreedyActionSelector(cfg.algorithm.epsilon)
        # self.train_policy = Agents(critic, explorer, PrintAgent())
        self.train_policy = Agents(critic, explorer)

        self.eval_policy = Agents(critic, ArgmaxActionSelector())

        # Creates two temporal agents just for "replaying" some parts
        # of the transition buffer
        self.t_q_agent = TemporalAgent(critic)
        self.t_target_q_agent = TemporalAgent(target_critic)

        # Get an agent that is executed on a complete workspace
        self.optimizer = setup_optimizer(cfg.optimizer, self.t_q_agent)

        self.last_critic_update_step = 0

def run(dqn: DQN, compute_critic_loss):
    for rb in dqn.iter_replay_buffers():
        for _ in range(dqn.cfg.algorithm.n_updates):
            rb_workspace = rb.get_shuffled(dqn.cfg.algorithm.batch_size)

            # The q agent needs to be executed on the rb_workspace workspace
            dqn.t_q_agent(rb_workspace, t=0, n_steps=2)
            with torch.no_grad():
                dqn.t_target_q_agent(rb_workspace, t=0, n_steps=2)

            q_values, terminated, reward, action, target_q_values = rb_workspace[
                "q_values", "env/terminated", "env/reward", "action", "target/q_values"
            ]

            # Determines whether values of the critic should be propagated
            must_bootstrap = ~terminated

            # Compute critic loss
            critic_loss = compute_critic_loss(
                dqn.cfg, reward, must_bootstrap, q_values, target_q_values, action
            )
            # Store the loss for tensorboard display
            dqn.logger.add_log("critic_loss", critic_loss, dqn.nb_steps)
            dqn.logger.add_log("q_values/min", q_values.max(-1).values.min(), dqn.nb_steps)
            dqn.logger.add_log("q_values/max", q_values.max(-1).values.max(), dqn.nb_steps)
            dqn.logger.add_log("q_values/mean", q_values.max(-1).values.mean(), dqn.nb_steps)

            dqn.optimizer.zero_grad()
            critic_loss.backward()
            torch.nn.utils.clip_grad_norm_(
                dqn.t_q_agent.parameters(), dqn.cfg.algorithm.max_grad_norm
            )
            dqn.optimizer.step()

            # Update target
            if (
                dqn.nb_steps - dqn.last_critic_update_step
                > dqn.cfg.algorithm.target_critic_update
            ):
                dqn.last_critic_update_step = dqn.nb_steps
                copy_parameters(dqn.t_q_agent, dqn.t_target_q_agent)

            dqn.evaluate()
```

```python
class FeatureFilterWrapper(gym.Wrapper):
    def __init__(self, env, filter):
        super().__init__(env)
        self.filter = filter
        self.observation_space = gym.spaces.Box(
            low=np.delete(env.observation_space.low, filter),
            high=np.delete(env.observation_space.high, filter),
            dtype=np.float32
        )
        # print(self.observation_space)

    def reset(self, **kwargs):
        obs, _ = self.env.reset(**kwargs)
        return np.delete(obs, self.filter), _

    def step(self, action):
        obs, reward, done, truncated, info = self.env.step(action)
        return np.delete(obs, self.filter), reward, done, truncated, info
```

```python
class ObsTimeExtensionWrapper(gym.Wrapper):
    def __init__(self, env):
        super().__init__(env)
        self.observation_space = gym.spaces.Box(
            low=np.concatenate((env.observation_space.low, env.observation_space.low), axis=0),
            high=np.concatenate((env.observation_space.high, env.observation_space.high), axis=0),
            dtype=np.float32
        )
        # print(np.concatenate((env.observation_space.low, env.observation_space.low), axis=0))
        self.memory = np.zeros(self.env.observation_space.shape[0])
        # print(self.observation_space)
        # print(self.memory.shape)

    def step(self, action):
        # print(action)
        obs, reward, done, truncated, info = self.env.step(action)
        # print(obs)
        # print(self.memory)
        observation = np.concatenate((obs, self.memory))
        self.memory = obs
        return observation, reward, done, truncated, info

    def reset(self, **kwargs):
        # print("here!")
        obs, _ = self.env.reset(**kwargs)
        self.memory = np.zeros(self.env.observation_space.shape[0])
        # print(self.memory.shape)
        observation = np.concatenate((obs, self.memory))
        return observation, _
```

```python
class FeatureObsTimeExtensionWrapper(gym.Wrapper):
    def __init__(self, env, features = []):
        super().__init__(env)
        self.features = features
        self.memory = np.zeros(len(features))

        self.observation_space = gym.spaces.Box(
            low=np.concatenate((env.observation_space.low, env.observation_space.low[features]), axis=0),
            high=np.concatenate((env.observation_space.high, env.observation_space.high[features]), axis=0),
            dtype=np.float32
        )
        # print(self.observation_space)
        # print(self.memory.shape)


    def observation(self, obs):
        return np.append(obs, self.memory)

    def step(self, action):
        obs, reward, terminated, truncated, info = self.env.step(action)
        output = self.observation(obs), reward, terminated, truncated, info
        self.memory = obs[self.features]
        return output

    def reset(self, **kwargs):
        obs, a = self.env.reset(**kwargs)
        self.memory = self.memory = np.zeros(len(self.features))
        return self.observation(obs), a
```

```python
class ActionTimeExtensionWrapper(gym.Wrapper):
    def __init__(self, env, nb_repeats):
        super().__init__(env)
        self.nb_repeats = nb_repeats
        self.action_space = gym.spaces.Discrete(int(math.pow(env.action_space.n, self.nb_repeats)))
        self.code = combinaisons_successives(env.action_space.n, self.nb_repeats)

    def step(self, action):
        # print(action)
        action_code = self.code[action]
        # print(action_code)
        valid_action = action_code[0].item()
        # print(valid_action)
        obs, reward, done, truncated, info = self.env.step(valid_action)
        return obs, reward, done, truncated, info

    def reset(self, **kwargs):
        obs, _ = self.env.reset(**kwargs)
        return obs, _
```
